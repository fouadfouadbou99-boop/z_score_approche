### Automating Calculations for GitHub and Streamlit

To prepare your Python calculations for GitHub and Streamlit, it's best to encapsulate them within functions. This makes your code modular, reusable, and easier to integrate into web applications like Streamlit.

````python
def calculate_sum(a, b):
    """Calculates the sum of two numbers."""
    return a + b

def calculate_product(a, b):
    """Calculates the product of two numbers."""
    return a * b

print("Functions for calculation defined.")
````

#### Demonstration of Usage

Let's see how to use these functions:

````python
num1 = 10
num2 = 5

sum_result = calculate_sum(num1, num2)
product_result = calculate_product(num1, num2)

print(f"The sum of {num1} and {num2} is: {sum_result}")
print(f"The product of {num1} and {num2} is: {product_result}")
````

````python
%%writefile calculations.py

import pandas as pd

def calculate_sum(a, b):
    """Calculates the sum of two numbers."""
    return a + b

def calculate_product(a, b):
    """Calculates the product of two numbers."""
    return a * b

def read_excel_file(file_path):
    """Reads an Excel file into a pandas DataFrame."""
    try:
        df = pd.read_excel(file_path)
        return df
    except Exception as e:
        return f"Erreur lors de la lecture du fichier Excel: {e}"

# Vous pouvez ajouter d'autres fonctions de calcul ici
````

### Fichier 2: `app.py`

Ce fichier sera votre application Streamlit. Il importera les fonctions de `calculations.py`.
Pour le télécharger :
1.  Cliquez sur le menu `Fichier` (File) en haut de Colab.
2.  Sélectionnez `Télécharger` (Download).
3.  Choisissez `Télécharger en tant que script Python (.py)` (Download as Python script (.py)).
4.  Renommez le fichier téléchargé en `app.py`.

````python
%%writefile app.py
import streamlit as st
from calculations import calculate_sum, calculate_product, read_excel_file # Importe les fonctions de votre fichier calculations.py
import pandas as pd # Importe pandas pour l'affichage des DataFrames

st.title('Application de Calcul Simple et Lecteur Excel')
st.write("Entrez deux nombres pour effectuer des calculs ou téléchargez un fichier Excel.")

# Section pour les calculs
st.header('Calculatrice de Nombres')
num1 = st.number_input('Premier nombre', value=0.0, key='num1_calc')
num2 = st.number_input('Deuxième nombre', value=0.0, key='num2_calc')

col1, col2 = st.columns(2)
with col1:
    if st.button('Calculer la Somme'):
        sum_result = calculate_sum(num1, num2)
        st.success(f'La somme de {num1} et {num2} est : {sum_result}')
with col2:
    if st.button('Calculer le Produit'):
        product_result = calculate_product(num1, num2)
        st.success(f'Le produit de {num1} et {num2} est : {product_result}')

# Section pour le lecteur Excel
st.header('Lecteur de Fichier Excel')
uploaded_file = st.file_uploader("Téléchargez votre fichier Excel (.xlsx)", type=["xlsx"])

if uploaded_file is not None:
    # Pour lire le fichier directement à partir du contenu uploadé
    # Streamlit gère le fichier comme un objet BytesIO
    df = read_excel_file(uploaded_file)
    if isinstance(df, pd.DataFrame):
        st.write("Contenu du fichier Excel:")
        st.dataframe(df)
    else:
        st.error(df)
````
